```python
import numpy as np
from scipy import stats

# ------------------------------------------------------------------
# 1. Dataset Generation (Simulating Normal Distributions)
# ------------------------------------------------------------------
np.random.seed(42)  # Set seed to ensure reproducible results across runs

# Experimental parameters
n1, n2 = 30, 30      # Sample sizes for both groups
mu1, mu2 = 105, 98   # True population means
sigma = 10           # Shared standard deviation (assumes equal population variances)

# Draw random samples from normal distributions
group1 = np.random.normal(loc=mu1, scale=sigma, size=n1)
group2 = np.random.normal(loc=mu2, scale=sigma, size=n2)

# ------------------------------------------------------------------
# 2. Descriptive Statistics & Mean Difference
# ------------------------------------------------------------------
# Compute sample means
mean1 = np.mean(group1)
mean2 = np.mean(group2)

# Compute unbiased sample variances (using N-1 degrees of freedom)
var1 = np.var(group1, ddof=1)
var2 = np.var(group2, ddof=1)

# Absolute difference between sample means
mean_diff = mean1 - mean2

# ------------------------------------------------------------------
# 3. Inferential Calculations (Pooled Variance & t-Test)
# ------------------------------------------------------------------
# Total degrees of freedom for pooled two-sample t-test
df = n1 + n2 - 2

# Calculate pooled sample variance: s_p^2 = [ (n1 - 1)*s1^2 + (n2 - 1)*s2^2 ] / df
s_p_sq = ((n1 - 1) * var1 + (n2 - 1) * var2) / df

# Calculate standard error of the mean difference
se_diff = np.sqrt(s_p_sq * (1/n1 + 1/n2))

# Calculate test statistic
t_stat = mean_diff / se_diff

# Calculate two-tailed p-value using the Student's t cumulative distribution function
p_val = 2 * (1 - stats.t.cdf(np.abs(t_stat), df=df))

# ------------------------------------------------------------------
# 4. 95% Confidence Interval Estimation
# ------------------------------------------------------------------
confidence_level = 0.95
alpha = 1 - confidence_level

# Two-tailed critical t-value for alpha = 0.05
t_critical = stats.t.ppf(1 - alpha/2, df=df)

# Calculate boundary bounds for the confidence interval
margin_of_error = t_critical * se_diff
ci_lower = mean_diff - margin_of_error
ci_upper = mean_diff + margin_of_error

# ------------------------------------------------------------------
# 5. Output Results & Statistical Decision
# ------------------------------------------------------------------
print("==================================================")
print("1. HYPOTHESES")
print("==================================================")
print("H0: mu1 - mu2 = 0 (There is no difference between the population means)")
print("H1: mu1 - mu2 != 0 (There is a significant difference between the population means)\n")

print("==================================================")
print("2. DESCRIPTIVE STATISTICS")
print("==================================================")
print(f"Group 1 Mean:              {mean1:.4f}")
print(f"Group 2 Mean:              {mean2:.4f}")
print(f"Group 1 Variance (ddof=1): {var1:.4f}")
print(f"Group 2 Variance (ddof=1): {var2:.4f}")
print(f"Mean Difference:           {mean_diff:.4f}\n")

print("============")
print("3. INFERENTIAL STATISTICS")
print("============")
print(f"Pooled Variance (s_p^2):    {s_p_sq:.4f}")
print(f"Degrees of Freedom (df):   {df}")
print(f"t-statistic:               {t_stat:.4f}")
print(f"p-value (two-tailed):      {p_val:.4e}\n")

print("==================")
print("4. 95% CONFIDENCE INTERVAL")
print("==================================================")
print(f"Margin of Error:           {margin_of_error:.4f}")
print(f"95% CI for (mu1 - mu2):   ({ci_lower:.4f}, {ci_upper:.4f})\n")

print("==================================================")
print("5. DECISION & INTERPRETATION")
print("==================================================")
if p_val < alpha:
    print(f"Decision: Reject H0 at alpha = {alpha}.")
    print(f"Interpretation: There is statistically significant evidence that the two population means are different (t({df}) = {t_stat:.2f}, p = {p_val:.4f}). On average, Group 1 exceeds Group 2 by {mean_diff:.2f} units.")
else:
    print(f"Decision: Fail to Reject H0 at alpha = {alpha}.")
    print(f"Interpretation: There is insufficient evidence to demonstrate a significant difference between the two population means (t({df}) = {t_stat:.2f}, p = {p_val:.4f}).")
```
